# Classifying images

```python
from hydrotools import classification

classification.classify_bathymetry(
    "example_01m.tif", "classified.jpg", method="quantile", n_classes=10
)

classification.extract_features("classified.jpg", "extracted_features.shp")

classification.rasterize_shapefile_outline(
    "extracted_features.shp", "extracted_features.png"
)
```
